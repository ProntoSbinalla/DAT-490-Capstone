# Cleaning Dataset

## Importing Libraries

```python
import pandas as pd
import numpy as np
```

## Importing Dataset

```python
data = pd.read_csv("../data/raw/full_dataset.csv")
geolocation = pd.read_csv("../data/raw/olist_geolocation_dataset.csv")

display(data.head())
display(geolocation.head())
```

## Merging Datasets with GeoLocation

```python
geo_avg = (
    geolocation.groupby("geolocation_zip_code_prefix")[["geolocation_lat", "geolocation_lng"]]
    .mean()
    .reset_index()
)

geo_customer = geo_avg.rename(columns={
    "geolocation_zip_code_prefix": "customer_zip_code_prefix",
    "geolocation_lat": "customer_lat",
    "geolocation_lng": "customer_lng"
})


data = data.merge(geo_customer, on="customer_zip_code_prefix", how="inner")

geo_seller = geo_avg.rename(columns={
    "geolocation_zip_code_prefix": "seller_zip_code_prefix",
    "geolocation_lat": "seller_lat",
    "geolocation_lng": "seller_lng"
})

data = data.merge(geo_seller, on="seller_zip_code_prefix", how="inner")

data.head()
```

```python
def haversine(lat1, lng1, lat2, lng2):
    # convert degrees to radians
    lat1, lng1, lat2, lng2 = map(np.radians, [lat1, lng1, lat2, lng2])
    
    dlat = lat2 - lat1
    dlng = lng2 - lng1
    
    a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlng/2)**2
    c = 2 * np.arcsin(np.sqrt(a))
    r = 6371  # Radius of Earth in kilometers
    return c * r

data["distance_km"] = haversine(
    data["customer_lat"], data["customer_lng"],
    data["seller_lat"], data["seller_lng"]
)
```

```python
display(data.columns)
display(data.shape)
```

## Pre-Processing Data Columns

### Removing Unnecessary Columns

```python
# Rename the incorrect column name
data.rename(columns={"product_name_lenght": "product_name_length",
                     'product_description_lenght': 'product_description_length',
                     }, inplace=True)

data.drop(columns=['review_comment_title',
                   'review_comment_message',
                   'review_creation_date',
                   'review_answer_timestamp',
                   'product_category_name',
                   'payment_sequential',
                   'payment_type',
                   'payment_installments',
                   'payment_value'], inplace=True)

display(data.columns)

display(data.shape)
```

### Removing Timestamps and Only Keeping Dates

```python
datetime_columns = [
    'order_purchase_timestamp', 'order_approved_at',
    'order_delivered_carrier_date', 'order_delivered_customer_date',
    'order_estimated_delivery_date', 'shipping_limit_date'
]

# Convert to datetime and strip time (keep only date)
for col in datetime_columns:
    if col in data.columns:
        data[col] = pd.to_datetime(data[col]).dt.date

# Display to verify changes
data.head()
```

### Creating Custom Columns

```python
# Convert both columns to datetime64[ns] before subtraction
data['handoff_delay'] = (
    pd.to_datetime(data['order_delivered_carrier_date']) -
    pd.to_datetime(data['shipping_limit_date'])
).dt.days

data['delivery_delay'] = (
    pd.to_datetime(data['order_delivered_customer_date']) -
    pd.to_datetime(data['order_estimated_delivery_date'])
).dt.days

# Delivered late remains a boolean comparison
data['delivered_late'] = (
    pd.to_datetime(data['order_delivered_customer_date']) >
    pd.to_datetime(data['order_estimated_delivery_date'])
)

# Preview the results
display(data.head())
```

```python
display(data.columns)
display(data.shape)
```

### Removing Duplicate Rows

```python
# Remove fully duplicated rows
data.drop_duplicates(inplace=True)

# Reset index after dropping duplicates
data.reset_index(drop=True, inplace=True)

# Display updated DataFrame
display(data.head())
display(data.shape)
display(data.columns)
```

## Re-Ordering Columns

```python
# Define the final desired column order based on your prior instructions and the current columns
final_column_order = [
    # Order and Customer Metadata
    'order_id', 'order_item_id', 'order_status', 'order_purchase_timestamp',
    'order_approved_at', 'order_estimated_delivery_date', 'is_canceled',
    
    # Customer Information
    'customer_id', 'customer_unique_id', 'customer_zip_code_prefix',
    'customer_city', 'customer_state', 'customer_lat', 'customer_lng',
    'order_delivered_customer_date', 'is_first_time_buyer',
    
    # Seller Information
    'seller_id', 'seller_zip_code_prefix', 'seller_city', 'seller_state',
    'seller_lat', 'seller_lng', 'shipping_limit_date',
    'order_delivered_carrier_date', 'handoff_delay',
    
    # Distance Between Seller and Customer
    'distance_km',
    
    # Product Information
    'product_id', 'product_category_name_english', 'price', 'freight_value',
    'product_name_length', 'product_description_length', 'product_photos_qty',
    'product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm',
    
    # Review Information
    'review_id', 'review_score',
    
    # Delivery Delay Information
    'delivered_late', 'delivery_delay'
]

# Reorder the columns
data = data[final_column_order]

display(data.head())
display(data.shape)
display(data.columns)
```

## Export to CSV

```python
data.to_csv("../data/cleaned/cleaned_dataset.csv", index=False)
```
